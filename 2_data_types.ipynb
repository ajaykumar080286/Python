**Python Data Types Categories**

***Basic Types***:

**Int**: Whole nums (1, 2, 3).<br>
**Float**: Decimals (3.14, 2.5).<br>
**Complex**: Real+Imag (1 + 2j).<br>
**Bool**: True/False.<br>
**Str**: Text ("hello").

**Container Types**:<br>
**List**: Ordered items.
[1, 2, 3]<br>
**Tuple**: Ordered, immutable.
(1, 2, 3)<br>
**Set**: Unordered, unique.
{1, 2, 3}<br>
**Dict**: Key-value pairs.
{'k1': 'v1', 'k2': 'v2'}<br>
**User-Defined Types:**
Class: Obj blueprint.
