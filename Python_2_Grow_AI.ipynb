**SET Operation**

```python
python_topic={
    "funaction",
    "list",
    "loop",
    "dictionary"

}

llm_topic={
    "funaction",
    "RAG",
    "Langchain",
    "langraph"

}

print("common",python_topic & llm_topic)
print("only in python",python_topic - llm_topic)
print("only in llm", llm_topic-python_topic)
print("all", llm_topic | python_topic)
```

**Decorator -** Adding Value into existing value without changing original  value

```python
def my_decorator(func):
    def wrapper():
        print("Before execution")
        func()
        print("After execution")
    return wrapper

@my_decorator
def greet():
    print("hello trainees")

greet()
```
